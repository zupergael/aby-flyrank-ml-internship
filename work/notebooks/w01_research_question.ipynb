## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import os, sys, subprocess
import pandas as pd
import numpy as np

# Automatically clone repo and change working directory if running in Google Colab
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("../..")

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

# Real Number 1: Total size and base rate
total_pages = len(df)
declining_rate = df["is_declining_label"].mean()

# Real Number 2: Baseline Precision@50
stale = (df["days_since_last_update"] >= 180).astype(int)
visible = (df["impressions_90d"] >= 500).astype(int)
df["hand_rule_score"] = stale * visible * df["impressions_90d"]

def precision_at_k(scores, labels, k=50):
    order = np.argsort(-np.asarray(scores))
    topk = np.asarray(labels)[order[:k]]
    return topk.mean()

p50_hand_rule = precision_at_k(df["hand_rule_score"], df["is_declining_label"], 50)

# Real Number 3: Correlation on active pages
active_pages = df[df["impressions_90d"] > 0]
corr_vol_imp = active_pages["search_volume"].corr(active_pages["impressions_90d"])

print(f"Real Number 1 -> Total Dataset: {total_pages:,} pages | Declining Rate: {declining_rate:.1%}")
print(f"Real Number 2 -> Baseline Hand Rule Precision@50: {p50_hand_rule:.3f}")
print(f"Real Number 3 -> Search Volume vs Traffic Correlation (active pages): {corr_vol_imp:.3f}")
```
